```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = df[(df["price"] < lower_bound) | (df["price"] > upper_bound)]

print("Number of price outliers:", len(outliers))
```

```python
df["price"] = df["price"].clip(upper=upper_bound)
```

```python
df["has_reviews"] = (df["number_of_reviews"] > 0).astype(int)

df["availability_category"] = pd.cut(
    df["availability_365"],
    bins=[-1, 90, 180, 365],
    labels=["Low", "Medium", "High"]
)

df["minimum_nights_category"] = pd.cut(
    df["minimum_nights"],
    bins=[0, 3, 7, 30, 365],
    labels=["Short Stay", "Weekly Stay", "Monthly Stay", "Long Stay"]
)

df["host_activity_level"] = pd.cut(
    df["calculated_host_listings_count"],
    bins=[0, 1, 5, 10, float("inf")],
    labels=["Single Listing", "Small Host", "Medium Host", "Large Host"]
)

df.head()
```
